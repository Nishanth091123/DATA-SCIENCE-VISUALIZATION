## Descriptive Analysis & Outlier Treatment

```python
# Load the user defined function

from Descriptive import Descriptive

obj=Descriptive()
```

```python
#Split Quantative and Qualtative column names
quantative,qualtative=obj.segreQuanQual(dataset)
```

```python
#Descriptive Porcess
quantative = dataset.select_dtypes(include='number').columns.tolist()
print("Quantitative Columns:", quantative)
```

```python
quantative = dataset.select_dtypes(include='number').columns.tolist()
print(quantative)
```

```python
des_Data = obj.descriptive_Analysis(dataset, quantative)
des_Data
```

```python
#Checking the outlier
lesser, greater = obj.outliercolumn(quantative, des_Data)

print("Lesser:", lesser)
print("Greater:", greater)
```

```python
# if outlier present run this, to replace the value
change_Data=obj.changeoutlier(dataset,des_Data,lesser,greater)
change_Data
```

```python
# After replacement of outlier we are checking with Descriptive analysis
des_Data=obj.descriptive_Analysis(dataset,quantative)
des_Data
```

# 1. Univariate Analysis

```python
# Histogram + KDE (modern replacement for distplot)
sb.histplot(dataset['ssc_p'], kde=True, color='steelblue')
plt.title('SSC % - Histogram with KDE')
plt.show()
```

```python
# KDE Plot
sb.kdeplot(data=dataset, x='mba_p', fill=True, color='red')
plt.title('MBA % - KDE Plot')
plt.show()
```

```python
# KDE Plot with hue
sb.kdeplot(data=dataset, x='etest_p', hue='status', fill=True)
plt.title('E-test % by Placement Status')
plt.show()
```

```python
# Displot (figure-level distribution plot)
sb.displot(data=dataset, x='degree_p', hue='gender', kde=True, col='status')
plt.show()
```

```python
# ECDF Plot
sb.ecdfplot(data=dataset, x='hsc_p', hue='gender')
plt.title('HSC % - ECDF Plot')
plt.show()
```

```python
# Rug Plot with KDE
sb.kdeplot(data=dataset, x='salary')
sb.rugplot(data=dataset, x='salary', color='black')
plt.title('Salary - KDE with Rug Plot')
plt.show()
```

```python
# Count Plot (single categorical column)
sb.countplot(x='status', data=dataset)
plt.title('Placement Status Count')
plt.show()
```

```python
# Count Plot - HSC Stream
sb.countplot(x='hsc_s', data=dataset, order=dataset['hsc_s'].value_counts().index)
plt.title('HSC Stream Count')
plt.show()
```

# 9. Matrix Plot

```python
# Correlation Heatmap
corr = dataset[['ssc_p','hsc_p','degree_p','etest_p','mba_p','salary']].corr()
sb.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f')
plt.title('Correlation Heatmap')
plt.show()
```

```python
# Heatmap - Crosstab of categories
ct = pd.crosstab(dataset['specialisation'], dataset['status'])
sb.heatmap(ct, annot=True, fmt='d', cmap='Blues')
plt.title('Specialisation vs Status')
plt.show()
```

```python
# Cluster Map
sb.clustermap(corr, annot=True, cmap='viridis')
plt.show()
```
